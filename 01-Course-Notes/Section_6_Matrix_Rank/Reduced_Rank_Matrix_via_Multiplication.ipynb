```python
# ======================================================================
# SECTION: REDUCED-RANK MATRIX GENERATION
# COMPLEXITY: O(m * n * r) - Matrix multiplication of (m x r) and (r x n).
# DATA SHAPE: Left (m x r) @ Right (r x n) -> Result (m x n).
# MATH: rank(A @ B) <= min(rank(A), rank(B))
# TOOLS: np.random.randn(), @ (matmul), np.linalg.matrix_rank()
# ======================================================================

import numpy as np

# set global print options
np.set_printoptions(suppress=True, precision=3)
```

# Reduced-Rank Matrix Generation via Multiplication

This module demonstrates the synthesis of low-rank matrices through the multiplication of non-square "tall" and "wide" factors. By defining a target rank $r$ that is lower than the dimensions of the final $m \times n$ matrix, we effectively compress the information space and force linear dependencies.

### Key Concepts
* **Rank Property:** The rank of the product of two matrices is limited by the rank of the factors. By multiplying an $m \times r$ and an $r \times n$ matrix, the resulting $m \times n$ matrix will have a maximum possible rank of $r$.
* **AI Application:** This is the fundamental logic behind **Matrix Factorization** and **Low-Rank Approximation**, used in recommendation engines and data compression to filter noise and identify latent features.

```python
def create_reduced_rank_matrix(row_count: int, column_count: int, target_rank: int) -> np.ndarray:
    """Create a row_count x column_count matrix with rank at most target_rank."""
    left_factor = np.random.randn(row_count, target_rank)
    right_factor = np.random.randn(target_rank, column_count)
    return left_factor @ right_factor


def print_matrix_summary(matrix: np.ndarray) -> None:
    """Print the shape and numerical rank of a matrix."""
    print(f"Matrix shape: {matrix.shape}")
    print(f"Matrix rank: {np.linalg.matrix_rank(matrix)}\n")

```

```python
# Create a 10 x 10, rank-4 matrix using matrix multiplication.
SQUARE_MATRIX_ROWS = 10
SQUARE_MATRIX_COLUMNS = 10
SQUARE_MATRIX_RANK = 4

reduced_rank_10x10 = create_reduced_rank_matrix(
    SQUARE_MATRIX_ROWS,
    SQUARE_MATRIX_COLUMNS,
    SQUARE_MATRIX_RANK,
)

print_matrix_summary(reduced_rank_10x10)

# Generalize the implementation to an m x n, rank-r matrix.
GENERAL_MATRIX_ROWS = 9
GENERAL_MATRIX_COLUMNS = 85
GENERAL_MATRIX_RANK = 6

reduced_rank_9x85 = create_reduced_rank_matrix(
    GENERAL_MATRIX_ROWS,
    GENERAL_MATRIX_COLUMNS,
    GENERAL_MATRIX_RANK,
)

print_matrix_summary(reduced_rank_9x85)
```
