## 3. Making a Matrix Full-Rank via "Shifting"
Based on [Lecture 69: Making a matrix full-rank by "shifting"](https://wgu.udemy.com/course/linear-algebra-theory-and-implementation/learn/lecture/10500648#overview), this technique addresses the "numerical ghosts" of singular matrices.

### The Problem
In computational linear algebra, a **reduced-rank (singular)** matrix is problematic because it has no inverse ($A^{-1}$ does not exist). This causes many algorithms, such as the Normal Equations in Linear Regression, to fail or return "NaN" results.

### The Solution: Regularization
By adding a tiny amount of information to the diagonal (where the eigenvalues live), we can "nudge" a singular matrix into becoming a full-rank, invertible matrix.

### The Formula
$$A_{shifted} = A + \lambda I$$

* **$A$**: The original $m \times m$ reduced-rank matrix.
* **$\lambda$ (Lambda)**: A tiny scalar (e.g., $0.001$), often called **epsilon** or the **regularization parameter**.
* **$I$**: The identity matrix of the same dimensions as $A$.

> **Key Takeaway:** Shifting artificially inflates the rank to be full-rank ($m$) without significantly altering the underlying data structure. In data science, this is the foundation of **Ridge Regression** (Tikhonov regularization).

```python
MATRIX_SHAPE = 40

def create_rank_deficient_matrix(shape: int | tuple[int, int]) -> np.ndarray:
    """
    Create a random matrix with a rank deficiency of 1.
    """
    if isinstance(shape, int):
        rows, cols = shape, shape
    else:
        rows, cols = shape

    if rows < 2:
        raise ValueError("Matrix must have at least two rows to create a dependent row.")

    def_matrix = np.random.randn(rows, cols)

    row_indices = list(range(rows))
    row_to_copy_from, row_to_overwrite = random.sample(row_indices, 2)
    def_matrix[row_to_overwrite, :] = def_matrix[row_to_copy_from, :]

    return def_matrix

def print_matrix_rank(target_matrix: np.ndarray, label: str) -> None:
    """Calculates and prints the rank of a given matrix."""
    matrix_rank = np.linalg.matrix_rank(target_matrix)
    print(f"Rank of {label}: {matrix_rank}")

def main():
    """Main execution function to demonstrate matrix shifting."""
    matrix = create_rank_deficient_matrix(MATRIX_SHAPE)
    print_matrix_rank(matrix, "Original Matrix")

    if matrix.shape[0] != matrix.shape[1]:
        raise ValueError("Matrix shifting requires a square matrix.")

    REGULARIZATION_STRENGTH = 0.001
    shifted_matrix = matrix + REGULARIZATION_STRENGTH * np.eye(matrix.shape[0])
    print_matrix_rank(shifted_matrix, "Shifted Matrix")

if __name__ == "__main__":
    main()
```
